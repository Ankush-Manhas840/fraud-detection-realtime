```python
import sys
sys.path.append("..")
import pandas as pd
from src.data import load_raw, clean
from src.features import build_features
from src.streaming.state import empty_card_memory, features_from_memory, update_memory

df = build_features(clean(load_raw("../data/raw/card_transaction.v1.csv", nrows=20000)))
```

```python
one_card = df[(df["user"] == 0) & (df["card"] == 0)]
len(one_card)
```

```python
bag = empty_card_memory()
live_rows = []

for _, purchase in one_card.iterrows():
    features = features_from_memory(bag, purchase)
    update_memory(bag, purchase)
    live_rows.append(features)

live = pd.DataFrame(live_rows)
```

```python
for col in ["new_shop", "new_shop_type", "gap_minutes", "count_last_hour", "amount_vs_usual", "share_same_hour"]:
    train_values = one_card[col].values
    live_values = live[col].astype(float).values
    same = ((train_values == live_values) | (pd.isna(train_values) & pd.isna(live_values))).mean()
    print(col, f"{same:.1%} match")
```
